```python
import pandas as pd
import statistics as sts
import os

# Define paths
user = '/Users/nancyrodas/Library/'
box = user + 'CloudStorage/Box-Box/'
project = box + 'NancyRodasDeLeon_Research/Projects/Bilingualism_VisualOddball_P3b/'
osf = project+ 'OSF/March2025/'
erps    = project + 'Data/EEG/erps/'
folder  = project + 'Data/EEG/Files/'

# Change working directory to project folder
os.chdir(osf)

# Read CSV file into DataFrame
nombre_file = pd.read_csv('DegreeofBilingualismP3b_FinalDataset.csv')

# Filter the data based on exclusion criteria
cocoa_df = nombre_file.loc[
    (nombre_file['Exclude'] == 0)]

# Sort DataFrame by composite score (from smallest to largest)
cocoa_df_sorted = cocoa_df.sort_values("Composite_Factor_Score").reset_index(drop=True)

# Calculate mean and standard deviation of the composite scores
composite_scores = cocoa_df_sorted['Composite_Factor_Score']
mean_composite  = sts.mean(composite_scores)       # e.g., 3.85
std_composite   = sts.stdev(composite_scores)        # e.g., 5.45

# Define threshold values based on ±2SD
one_std_below  = mean_composite - std_composite      # e.g., -1.60
one_std_above  = mean_composite + std_composite      # e.g., 9.30
two_std_below  = mean_composite - 2 * std_composite    # e.g., -7.05
two_std_above  = mean_composite + 2 * std_composite    # e.g., 14.75

# Create groups based on composite score thresholds
group1 = cocoa_df_sorted[cocoa_df_sorted['Composite_Factor_Score'] < one_std_below]
group2 = cocoa_df_sorted[(cocoa_df_sorted['Composite_Factor_Score'] >= one_std_below) &
                         (cocoa_df_sorted['Composite_Factor_Score'] <= mean_composite)]
group3 = cocoa_df_sorted[(cocoa_df_sorted['Composite_Factor_Score'] > mean_composite) &
                         (cocoa_df_sorted['Composite_Factor_Score'] <= one_std_above)]
group4 = cocoa_df_sorted[(cocoa_df_sorted['Composite_Factor_Score'] > one_std_above) &
                         (cocoa_df_sorted['Composite_Factor_Score'] <= two_std_above)]

# Display group sizes (for sanity check)
print("Group sizes:")
print("Group 1:", group1.shape[0])
print("Group 2:", group2.shape[0])
print("Group 3:", group3.shape[0])
print("Group 4:", group4.shape[0])

# Dictionary to loop through groups
groups = {'Group1': group1, 'Group2': group2, 'Group3': group3, 'Group4': group4}

# Loop through each group and write ERP file paths to a .txt file
for group_name, group_data in groups.items():
    output_file_path = os.path.join(folder, f"{group_name}_ByStandDev_Composite_Bling.txt")
    with open(output_file_path, "w") as group_file:
        # For each participant in the group, construct the ERP file path and write it
        for ID in group_data["ID"].astype(str):
            path = erps + ID + "_VO.erp"
            group_file.write(path + "\n")
            print(f"Path written to {output_file_path}: {path}")
    print(f"Finished writing paths to {output_file_path}")

```

```python
import pandas as pd
import os

# Define paths
user = '/Users/nancyrodas/Library/'
box = user + 'CloudStorage/Box-Box/'
project = box + 'NancyRodasDeLeon_Research/Projects/Bilingualism_VisualOddball_P3b/'
erps = project + 'Data/EEG/erps/'
folder = project + 'Data/EEG/Files/'

# Change working directory to project folder
os.chdir(project)

# Read CSV file into DataFrame
nombre_file = pd.read_csv('Bling_COCOA_comprehensive_dataset_IDEA_FinalDataset.csv')

# Filter the data based on exclusion criteria
cocoa_df = nombre_file.loc[
    (nombre_file['Visual_Inspection'] == 0) &
    (nombre_file['DemoExclude'] == 0) &
    (nombre_file['Behavior_exclude'] == 0)
]

# Sort DataFrame by composite score (from smallest to largest)
cocoa_df_sorted = cocoa_df.sort_values("Composite_Factor_Score").reset_index(drop=True)

# Create a new column "group" that assigns each participant to one of 7 groups (10 per group)
# This assumes you have exactly 70 participants after filtering.
cocoa_df_sorted["group"] = (cocoa_df_sorted.index // 10) + 1

# Loop through each group and write the corresponding ERP file paths to a separate .txt file
for group_num, group_data in cocoa_df_sorted.groupby("group"):
    output_file_path = folder + f"Group{group_num}_Composite_Bling.txt"
    with open(output_file_path, "w") as group_file:
        # Iterate over each participant in this group
        for ID in group_data["ID"].astype(str):
            # Construct the full ERP file path
            path = erps + ID + "_VO.erp"
            group_file.write(path + "\n")
            print(f"Path written to {output_file_path}: {path}")
    print(f"Finished writing paths to {output_file_path}")

```

```python
import pandas as pd
import statistics as sts
import os

# Define paths
user    = '/Users/nancyrodas/Library/'
box     = user + 'CloudStorage/Box-Box/'
project = box + 'NancyRodasDeLeon_Research/Projects/Bilingualism_VisualOddball_P3b/'
erps    = project + 'Data/EEG/erps/'
folder  = project + 'Data/EEG/Files/'

# Change working directory to project folder
os.chdir(project)

# Read CSV file into DataFrame
nombre_file = pd.read_csv('Bling_COCOA_comprehensive_dataset_IDEA_FinalDataset.csv')

# Filter the data based on exclusion criteria
cocoa_df = nombre_file.loc[
    (nombre_file['Visual_Inspection'] == 0) &
    (nombre_file['DemoExclude'] == 0) &
    (nombre_file['Behavior_exclude'] == 0)
]

# Sort DataFrame by composite score (from smallest to largest)
cocoa_df_sorted = cocoa_df.sort_values("Composite_Factor_Score").reset_index(drop=True)

# Calculate mean and standard deviation of the composite scores
composite_scores = cocoa_df_sorted['Composite_Factor_Score']
mean_composite  = sts.mean(composite_scores)
std_composite   = sts.stdev(composite_scores)

# Define threshold values based on ±2SD
one_std_below  = mean_composite - std_composite
one_std_above  = mean_composite + std_composite
two_std_below  = mean_composite - 2 * std_composite
two_std_above  = mean_composite + 2 * std_composite

# Create groups based on composite score thresholds:
# Group 1: Scores below one SD below the mean
group1 = cocoa_df_sorted[cocoa_df_sorted['Composite_Factor_Score'] < one_std_below]

# Group 2: Scores between one SD below the mean and the mean
group2 = cocoa_df_sorted[(cocoa_df_sorted['Composite_Factor_Score'] >= one_std_below) &
                         (cocoa_df_sorted['Composite_Factor_Score'] <= mean_composite)]

# Instead of one group for scores between the mean and one SD above,
# split this group (which has 26 participants) into two via a median split:
group3_full = cocoa_df_sorted[(cocoa_df_sorted['Composite_Factor_Score'] > mean_composite) &
                              (cocoa_df_sorted['Composite_Factor_Score'] <= one_std_above)]
median_group3 = group3_full['Composite_Factor_Score'].median()
group3 = group3_full[group3_full['Composite_Factor_Score'] <= median_group3]
group4 = group3_full[group3_full['Composite_Factor_Score'] > median_group3]

# Group 5: Scores above one SD above the mean up to two SD above
group5 = cocoa_df_sorted[(cocoa_df_sorted['Composite_Factor_Score'] > one_std_above) &
                         (cocoa_df_sorted['Composite_Factor_Score'] <= two_std_above)]

# Display group sizes (for sanity check)
print("Group sizes:")
print("Group 1:", group1.shape[0])
print("Group 2:", group2.shape[0])
print("Group 3:", group3.shape[0])
print("Group 4:", group4.shape[0])
print("Group 5:", group5.shape[0])

# Dictionary to loop through groups
groups = {
    'Group1': group1,
    'Group2': group2,
    'Group3': group3,
    'Group4': group4,
    'Group5': group5
}

# Loop through each group and write ERP file paths to a .txt file
for group_name, group_data in groups.items():
    output_file_path = os.path.join(folder, f"{group_name}_5GroupingSDs.txt")
    with open(output_file_path, "w") as group_file:
        # For each participant in the group, construct the ERP file path and write it
        for ID in group_data["ID"].astype(str):
            path = erps + ID + "_VO.erp"
            group_file.write(path + "\n")
            print(f"Path written to {output_file_path}: {path}")
    print(f"Finished writing paths to {output_file_path}")
```
