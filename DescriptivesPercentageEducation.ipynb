This script is for converting the labels into number categories to get min, max, mean, std or percent of participants

```python
import os
import pandas as pd


directory = '/Users/nancyrodas/Library/'
box = directory + 'CloudStorage/Box-Box/'
project = box + 'Word Memory/'
files = project + 'Excel_Files/'

os.chdir(files)
print(os.getcwd())

ses = pd.read_csv('PARENT_SES.csv')
# Load data into DataFrame
df = pd.DataFrame(ses)

# Filter string columns
string_columns = df.select_dtypes(include='object')

# Calculate percentages for each string column
percentages = {}
for column in string_columns:
    counts = string_columns[column].value_counts(normalize=True) * 100
    percentages[column] = counts

# Display percentages for each column
for col, counts in percentages.items():
    print(f"\nPercentages for column '{col}':")
    print(counts)

```
